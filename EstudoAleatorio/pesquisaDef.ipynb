```python
#!/usr/bin/env python3
# curso_curto_example.py
# Exemplos típicos que aparecem em um mini-curso de Python

import json
import os
import sys
from dataclasses import dataclass, asdict
from typing import List, Dict, Optional

# -------------------------
# 1) Variáveis, tipos e entrada/saída
# -------------------------
def exemplo_basico():
    nome = "Maria"
    idade = 28
    altura = 1.68
    ativo = True
    print(f"Nome: {nome}, Idade: {idade}, Altura: {altura}m, Ativo: {ativo}")

# -------------------------
# 2) Listas e dicionários
# -------------------------
def exemplo_collections():
    frutas = ["maçã", "banana", "laranja"]
    frutas.append("uva")
    print("Frutas:", frutas)

    pessoas = [
        {"nome": "Ana", "idade": 22},
        {"nome": "Bruno", "idade": 31},
    ]
    # Filtrar com list comprehension
    maiores_25 = [p for p in pessoas if p["idade"] > 25]
    print("Maiores de 25:", maiores_25)

# -------------------------
# 3) Funções (com docstring e typing)
# -------------------------
def soma(a: float, b: float) -> float:
    """Retorna a soma de dois números."""
    return a + b

def exemplo_funcoes():
    print("2 + 3 =", soma(2, 3))

# -------------------------
# 4) Manipulação de arquivos (JSON)
# -------------------------
DATA_FILE = "exemplo_dados.json"

def salvar_dados(dados: Dict):
    with open(DATA_FILE, "w", encoding="utf-8") as f:
        json.dump(dados, f, indent=2, ensure_ascii=False)
    print(f"Dados salvos em {DATA_FILE}")

def carregar_dados() -> Dict:
    if not os.path.exists(DATA_FILE):
        return {}
    with open(DATA_FILE, "r", encoding="utf-8") as f:
        return json.load(f)

def exemplo_arquivos():
    dados = {"usuarios": [{"nome": "Carlos", "idade": 40}]}
    salvar_dados(dados)
    lidos = carregar_dados()
    print("Dados lidos:", lidos)

# -------------------------
# 5) Classes (dataclass)
# -------------------------
@dataclass
class Usuario:
    nome: str
    idade: int
    email: Optional[str] = None

def exemplo_classes():
    u = Usuario(nome="João", idade=27, email="joao@example.com")
    print("Usuario:", asdict(u))

# -------------------------
# 6) Tratamento de erros
# -------------------------
def exemplo_erros(x: int):
    try:
        resultado = 10 / x
    except ZeroDivisionError:
        print("Erro: divisão por zero")
        return None
    except Exception as e:
        print("Outro erro:", e)
        return None
    else:
        print("Resultado:", resultado)
        return resultado
    finally:
        print("Bloco finally executado")

# -------------------------
# 7) Requisição HTTP simples (requests)
# -------------------------
# Observação: requests precisa estar instalado: pip install requests
def exemplo_requests():
    try:
        import requests
    except ImportError:
        print("Pacote 'requests' não instalado. Rode: pip install requests")
        return

    url = "https://api.github.com"  # exemplo público
    try:
        r = requests.get(url, timeout=5)
        print("Status:", r.status_code)
        js = r.json()
        print("Chaves recebidas do GitHub API:", list(js.keys())[:6])
    except requests.RequestException as e:
        print("Erro ao acessar API:", e)

# -------------------------
# 8) Pequeno CLI — gerenciador de contatos em memória
# -------------------------
CONTACTS_FILE = "contatos.json"

def load_contacts() -> List[Dict]:
    if os.path.exists(CONTACTS_FILE):
        with open(CONTACTS_FILE, "r", encoding="utf-8") as f:
            return json.load(f)
    return []

def save_contacts(contacts: List[Dict]):
    with open(CONTACTS_FILE, "w", encoding="utf-8") as f:
        json.dump(contacts, f, indent=2, ensure_ascii=False)

def add_contact(nome: str, telefone: str):
    contacts = load_contacts()
    contacts.append({"nome": nome, "telefone": telefone})
    save_contacts(contacts)
    print(f"Contato {nome} adicionado.")

def list_contacts():
    contacts = load_contacts()
    if not contacts:
        print("Sem contatos.")
        return
    for i, c in enumerate(contacts, 1):
        print(f"{i}. {c['nome']} — {c['telefone']}")

def cli():
    print("Mini-CLI de contatos. Comandos: add, list, exit")
    while True:
        cmd = input(">> ").strip().lower()
        if cmd == "add":
            nome = input("Nome: ").strip()
            tel = input("Telefone: ").strip()
            add_contact(nome, tel)
        elif cmd == "list":
            list_contacts()
        elif cmd in ("exit", "quit"):
            print("Saindo...")
            break
        else:
            print("Comando desconhecido. Tente: add, list, exit")

# -------------------------
# 9) Função main: demonstração rápida
# -------------------------
def main():
    print("== Exemplos do mini-curso ==")
    exemplo_basico()
    exemplo_collections()
    exemplo_funcoes()
    exemplo_classes()
    exemplo_arquivos()
    exemplo_erros(2)
    exemplo_erros(0)
    exemplo_requests()

    print("\nRodando a CLI de exemplo? (s/n)")
    escolha = input("Escolha: ").strip().lower()
    if escolha == "s":
        cli()
    else:
        print("Fim da execução. Arquivos criados:", DATA_FILE, CONTACTS_FILE)

if __name__ == "__main__":
    main()
```
